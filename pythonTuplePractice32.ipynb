# Exercise 2: The Trailing Comma
create a tuple containing a single item, the number 50, and confirm its type.

important: assuming that parentheses alone create a tuple is a mistake. A trailing comma is required when the tuple has only one element

```python
single_1 =(50)
print(f"single_1 : {single_1}  and {type(single_1)}")

single_one =(50,)
print(f"single_one : {single_one} and {type(single_one)}")
```
